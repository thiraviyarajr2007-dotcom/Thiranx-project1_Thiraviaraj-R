## step1:  Import  data-Load  the  Data

```python
import pandas as pd
```

```python
#Load the  dataset    

data=pd.read_csv("sample_data_cleaning_project - Sample_data_cleaning_project.csv")
print(data)
```

## step 2  : Identify  the    missing    values   using isnull().sum()  

```python
print("\nMissing Values Count:")
print(data.isnull().sum())
```

```python
data['Age']=data['Age'].fillna(data['Age'].median())
data['Salary']=data['Salary'].fillna(data['Salary'].mean())
data=data.dropna(subset=['Join_Date'])
print(data)
```

##  step  4  :  check and  remove  the  duplicate    rows   

```python
data=data.apply(lambda x:x.str.strip() if  x.dtype=="object" else x)
```

```python
data=data.apply(lambda x:x.str.lower() if  x.dtype=="object" else x)
print(data)
```

```python
data=data.drop_duplicates(subset=['Name','Age'])
print(data)             
```

```python
print(data.duplicated().sum())
```

## step 5 :   Handle    the    Outliers 

```python
# calculate IQR for Salary  

Q1=data['Salary'].quantile(0.25)
Q3=data['Salary'].quantile(0.75)
IQR=Q3-Q1

#  Filter the Outliers 

data = data[(data['Salary'] >= Q1 - 1.5 * IQR)   &  (data['Salary']<=Q3 + 1.5*IQR)]

print(data)
```

## Step 6:  Convert  Datatypes   

```python
data['Join_Date']=pd.to_datetime(data['Join_Date'])

print(data)
```

## Step 7 : Encode Categorical Variables 

```python

#  One -hot encode  the  'Department'  column
data=pd.get_dummies(data,columns=['Department'],drop_first=True)

#Convert only   one_hot   encoded columbs  to   integers  

binary_columns=[col for col in data.columns if  col.startswith('Department_')]
data[binary_columns]=data[binary_columns].astype(int)

print(data)
```

```python
data.to_csv("Cleaneddataset.csv", index=False)
```

##  Visuliazation

```python
import matplotlib.pyplot as plt

# load the  dataset 

data=pd.read_csv("Cleaneddataset.csv")

# Display  the  first few   rows of the  element  

print(data.head())
```

# Bar chart   for  departments  

```python
department_count=data[['Department_hr','Department_it']].sum()

# plating Department Distributions  

department_count.plot(kind='bar',color=['blue','green'],figsize=(8,6))
plt.title("Department Distribution")
plt.xlabel("Department")
plt.ylabel("Number of Employees")
plt.xticks(ticks=[0,1],labels=["HR","IT"],rotation=0)
plt.show()
```

## Histogram for   Salaries 

```python
plt.figure(figsize=(8,6))
plt.hist(data['Salary'],bins=10,color='orange',edgecolor='black')
plt.title("Salary Distribution")
plt.xlabel("Salary")
plt.ylabel("Frequency")
plt.grid(True)
plt.show()
```

##  Line plot for  Age  and  Salary

```python
plt.figure(figsize=(8,6))
plt.plot(data['Age'],data['Salary'],marker='o',linestyle='-',color='orange')
plt.title("Salary vs Age")
plt.xlabel("Age")
plt.ylabel("Salary")
plt.grid(True)
plt.show()
```

## Scatter plot  for   Salary   and  Department 

```python
plt.figure(figsize=(8,6))
colors=['blue' if dept ==1  else 'green' for dept in data['Department_hr']]
plt.scatter(data['Salary'],data['Age'],c=colors,alpha=0.7,edgecolor='black')
plt.title("Salary vs Age   by  Department (HR is  Blue &   IT in Green")
plt.ylabel("Age")
plt.xlabel("Salary")
plt.grid(True)
plt.show()
```
